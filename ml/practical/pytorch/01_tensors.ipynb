# 01 — Tensors

PyTorch's core data structure. Tensors are like NumPy arrays but support GPU acceleration and autograd.

## Creating Tensors

```python
import torch
import numpy as np

torch.tensor([1, 2, 3])
torch.tensor([[1., 2.], [3., 4.]])
torch.zeros(2, 3)
torch.ones(2, 3)
torch.full((2, 3), 7)
torch.eye(3)
torch.arange(0, 10, 2)
torch.linspace(0, 1, 5)
torch.rand(2, 3)          # uniform [0, 1)
torch.randn(2, 3)         # standard normal
torch.randint(0, 10, (2, 3))
torch.empty(2, 3)         # uninitialized memory
```

## dtype and device

```python
x = torch.tensor([1, 2, 3], dtype=torch.float32)
x.dtype
x.to(torch.int64)
x.float()
x.long()
x.double()

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
x = x.to(device)
x.device
```

## NumPy Interop

```python
arr = np.array([1, 2, 3])
t = torch.from_numpy(arr)      # shares memory with arr
t.numpy()                      # back to numpy (shares memory if on CPU)

# safe copy (no shared memory)
t2 = torch.tensor(arr)
```

## Shape, Indexing, Slicing

```python
x = torch.randn(4, 5)
x.shape
x.size()
x.dim()
x.numel()

x[0]            # first row
x[:, 1]         # second column
x[1:3, 2:4]     # sub-matrix
x[x > 0]        # boolean mask indexing

torch.masked_select(x, x > 0)
torch.where(x > 0, x, torch.zeros_like(x))
```

## Reshaping

```python
x = torch.arange(12)
x.view(3, 4)        # requires contiguous memory
x.reshape(3, 4)     # safer, copies if needed
x.view(-1, 4)       # infer dim

x = torch.randn(2, 3, 4)
x.permute(2, 0, 1).shape     # reorder dims
x.transpose(0, 1).shape      # swap two dims
x.T                          # 2D transpose shortcut

x = torch.randn(1, 3, 1)
x.squeeze().shape        # remove dims of size 1
x.squeeze(0).shape       # remove specific dim
x.unsqueeze(0).shape     # add a dim
```

## Combining Tensors

```python
a = torch.randn(2, 3)
b = torch.randn(2, 3)

torch.cat([a, b], dim=0)     # stack along existing dim -> (4, 3)
torch.cat([a, b], dim=1)     # -> (2, 6)
torch.stack([a, b], dim=0)   # new dim -> (2, 2, 3)
torch.split(a, 1, dim=0)     # split into chunks
torch.chunk(a, 2, dim=1)
```

## Math Operations & Broadcasting

```python
a = torch.randn(3, 1)
b = torch.randn(1, 4)
a + b                 # broadcasts to (3, 4)

x = torch.randn(3, 3)
x + 1
x * 2
x @ x                 # matrix multiply
torch.matmul(x, x)
x.sum()
x.sum(dim=0)
x.mean()
x.max(), x.min()
x.argmax(dim=1)
torch.exp(x)
torch.log(x.abs())
torch.clamp(x, min=-1, max=1)
```

## In-place Operations

Ops suffixed with `_` modify the tensor in place (saves memory, but breaks autograd history if misused).

```python
x = torch.randn(3)
x.add_(1)
x.mul_(2)
x.clamp_(min=0)
```

## Common Gotchas
- `view()` fails on non-contiguous tensors — use `.reshape()` or `.contiguous().view()`.
- In-place ops on tensors that require grad can break the autograd graph.
- Tensors on different devices can't be combined directly — always `.to(device)` first.
- `tensor.item()` extracts a Python scalar from a 1-element tensor.

```python
x = torch.tensor([3.14])
x.item()
```
